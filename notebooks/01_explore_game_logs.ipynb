```python
df['is_home'] = ~df['MATCHUP'].str.contains('@')
df['opponent'] = df['MATCHUP'].str.split().str[-1]
```

```python
df[['MATCHUP', 'is_home', 'opponent']].head()
```

```python
df = df.drop(columns=['MATCHUP'])
```

```python
from nba_api.stats.static import teams

all_teams = teams.get_teams()
team_lookup = {t['abbreviation']: t['id'] for t in all_teams}

df['opponent_team_id'] = df['opponent'].map(team_lookup)
```

```python
df.head()
```

```python
# Get current column order
cols = df.columns.tolist()

# Remove the ones we're relocating, and drop 'opponent' entirely
cols.remove('is_home')
cols.remove('opponent')
cols.remove('opponent_team_id')

# Find where GAME_DATE is, and insert the remaining two columns right after it
insert_at = cols.index('GAME_DATE') + 1
cols[insert_at:insert_at] = ['opponent_team_id', 'is_home']

# Apply the new order (opponent is now excluded)
df = df[cols]
```

```python
df = df.drop(columns=['FG_PCT', 'FG3_PCT', 'FT_PCT', 'REB', 'VIDEO_AVAILABLE'])
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)
df.head()
```
