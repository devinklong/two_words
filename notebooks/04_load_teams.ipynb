```python
import sys
sys.path.append('..')

from nba_api.stats.static import teams
from scripts.db_connection import get_connection
```

```python
all_teams = teams.get_teams()
conn = get_connection()
cur = conn.cursor()
```

```python
for t in all_teams:
    cur.execute(
        """INSERT INTO teams (team_id, abbreviation, full_name, city, state, year_founded)
           VALUES (%s, %s, %s, %s, %s, %s)""",
        (t['id'], t['abbreviation'], t['full_name'], t['city'], t['state'], t['year_founded'])
    )

conn.commit()
cur.close()
conn.close()
```
